# **Lab 11: Exploratory Data Analysis, Missing Data, and Outliers**
**AIM**

To identify missing values, outliers, numerical features, categorical features, and possible data-quality problems in the marketing campaign dataset.

**ALGORITHM**

1. Load the dataset.
2. Remove duplicates.
3. Find missing values.
4. Separate numerical and categorical attributes.
5. Calculate outliers using the IQR method.
6. Display numerical outliers using box plots.
7. Treat missing values.
8. Document the cleaned data.




```python
marketing_df = pd.read_csv("marketing_campaign.csv", sep=";")

print("Duplicate records:", marketing_df.duplicated().sum())

marketing_df = marketing_df.drop_duplicates().copy()

numerical_columns = marketing_df.select_dtypes(include=np.number).columns.tolist()
categorical_columns = marketing_df.select_dtypes(include=["object", "category", "bool"]).columns.tolist()

print("Numerical features:")
print(numerical_columns)

print("\nCategorical features:")
print(categorical_columns)

print("\nMissing-value summary:")
print(marketing_df.isnull().sum())

for column in numerical_columns:
    plt.figure(figsize=(6, 3))
    sns.boxplot(x=marketing_df[column])
    plt.title(f"Box Plot: {column}")
    plt.show()

outlier_summary = {}
for column in numerical_columns:
    q1 = marketing_df[column].quantile(0.25)
    q3 = marketing_df[column].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    outlier_summary[column] = (
        (marketing_df[column] < lower) | (marketing_df[column] > upper)
    ).sum()

print("Outlier summary:")
print(pd.Series(outlier_summary).sort_values(ascending=False))

for column in numerical_columns:
    marketing_df[column] = marketing_df[column].fillna(marketing_df[column].median())

for column in categorical_columns:
    marketing_df[column] = marketing_df[column].fillna(marketing_df[column].mode()[0])

print("Total missing values after treatment:", marketing_df.isnull().sum().sum())

marketing_df.to_csv("marketing_campaign_cleaned.csv", index=False)
```

# **Lab 13: Building Models for Marketing Campaign Response Prediction**
**AIM**

To build machine learning classification models that predict whether a customer will respond positively to a marketing campaign.

**ALGORITHM**

1. Load the cleaned marketing dataset.
2. Select the response column as the target.
3. Perform feature engineering.
4. Separate numerical and categorical variables.
5. Impute missing values.
6. Encode categorical variables.
7. Split data into training and testing sets.
8. Train classification models.
9. Generate predictions.



```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier

marketing_df = pd.read_csv("marketing_campaign_cleaned.csv")

marketing_df["TotalChildren"] = marketing_df["Kidhome"] + marketing_df["Teenhome"]
marketing_df["TotalSpending"] = marketing_df[spending_columns].sum(axis=1)
marketing_df["TotalPurchases"] = marketing_df[purchase_columns].sum(axis=1)

marketing_df = marketing_df.dropna(subset=[target_marketing])

X = marketing_df.drop(target_marketing, axis=1)
y = marketing_df[target_marketing]

if y.dtype == "object":
    y = y.astype("category").cat.codes

numeric_features = X.select_dtypes(include=["int64", "float64"]).columns.tolist()
categorical_features = X.select_dtypes(include=["object", "category", "bool"]).columns.tolist()

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore", sparse_output=False)),
])

preprocessor = ColumnTransformer([
    ("numerical", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features),
])

models = {
    "Logistic Regression": LogisticRegression(max_iter=2000, class_weight="balanced"),
    "Decision Tree": DecisionTreeClassifier(max_depth=8, random_state=42, class_weight="balanced"),
    "Random Forest": RandomForestClassifier(n_estimators=200, random_state=42, class_weight="balanced"),
    "Gradient Boosting": GradientBoostingClassifier(random_state=42),
}

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

marketing_predictions = {}

for name, classifier in models.items():
    model = Pipeline([("preprocessor", preprocessor), ("classifier", classifier)])
    model.fit(X_train, y_train)
    marketing_predictions[name] = model.predict(X_test)
    print(name, "training completed")
```

# **Lab 14: Accuracy and Comparative Analysis of Marketing Models**
**AIM**

To evaluate and compare the classification models used for marketing campaign response prediction.

**ALGORITHM**

1. Use the predictions obtained in Lab 13.
2. Calculate accuracy, precision, recall, and F1-score.
3. Generate confusion matrices.
4. Compare the models using a bar graph.
5. Select the most suitable model.




```python
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    classification_report, confusion_matrix,
)

marketing_results_df = pd.DataFrame([
    {
        "Model": name,
        "Accuracy": accuracy_score(y_test, prediction),
        "Precision": precision_score(y_test, prediction, average="weighted", zero_division=0),
        "Recall": recall_score(y_test, prediction, average="weighted", zero_division=0),
        "F1 Score": f1_score(y_test, prediction, average="weighted", zero_division=0),
    }
    for name, prediction in marketing_predictions.items()
])

display(marketing_results_df.sort_values("F1 Score", ascending=False))

plt.figure(figsize=(9, 5))
sns.barplot(data=marketing_results_df, x="Model", y="F1 Score", hue="Model", legend=False)
plt.ylim(0, 1)
plt.xticks(rotation=30)
plt.title("Marketing Response Model Comparison")
plt.show()

best_marketing_name = marketing_results_df.loc[marketing_results_df["F1 Score"].idxmax(), "Model"]
best_prediction = marketing_predictions[best_marketing_name]

cm = confusion_matrix(y_test, best_prediction)
plt.figure(figsize=(5, 4))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.title(f"Confusion Matrix: {best_marketing_name}")
plt.show()

print("Best model:", best_marketing_name)
print(classification_report(y_test, best_prediction, zero_division=0))
```
